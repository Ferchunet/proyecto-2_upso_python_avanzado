# Punto 4

MarieCurie.jpeg. Canny y Laplaciano.


```python
import cv2
from fotoapp import detectar_bordes_canny, detectar_bordes_laplaciano, mostrar
```

```python
img = cv2.imread("imgs/MarieCurie.jpeg")
# La convertimos a RGB porque matplotlib asume que la imagen es en RGB
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

img_bordes = detectar_bordes_canny(img)
cv2.imwrite("fotos_mejoradas/MarieCurie_Bordes_Canny.png", img_bordes)
mostrar([img, img_bordes], ["Original", "Bordes Canny"])
```

```python
img_bordes = detectar_bordes_laplaciano(img)
cv2.imwrite("fotos_mejoradas/MarieCurie_Bordes_Laplaciano.png", img_bordes)
mostrar([img, img_bordes], ["Original", "Bordes Laplaciano"])
```
